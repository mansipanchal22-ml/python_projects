# Pro-8 Numpy analyzer

```python
import numpy as np

class DataAnalytics:
    def __init__(self, data=None):
        #Encapsulation: Private attribute
        if data is not None:
            self.__array = np.array(data)
        else:
            self.__array = None

    #Static method utility
    @staticmethod
    def parse_input(input_str):
        """String input mathi space separate kari float array banave che."""
        return np.array([float(x) for x in input_str.strip().split()])

    # Class method utility
    @classmethod
    def from_list(cls, data_list):
        return cls(data_list)

    # Private helper method encapsulation mate
    def __has_array(self):
        return self.__array is not None

    def get_array(self):
        return self.__array

    # 1. Array Creation, Indexing, Slicing
    def create_array(self, dim, shape, elements):
        if dim == 1:
            self.__array = np.array(elements)
        elif dim == 2:
            self.__array = np.array(elements).reshape(shape[0], shape[1])
        elif dim == 3:
            self.__array = np.array(elements).reshape(shape[0], shape[1], shape[2])
        print("\nArray created successfully:")
        print(self.__array)

    def slice_array(self, row_start, row_end, col_start, col_end):
        if self.__has_array() and self.__array.ndim >= 2:
            sliced = self.__array[row_start:row_end, col_start:col_end]
            print("\nSliced Array:")
            print(sliced)
        else:
            print("\nArray 2D ke enathi vadhare hovud joie.")

    # 2. Mathematical Operations
    def math_operation(self, op_type, second_elements):
        if not self.__has_array():
            return
        second_array = np.array(second_elements).reshape(self.__array.shape)
        print("\nOriginal Array:\n", self.__array)
        print("Second Array:\n", second_array)

        if op_type == 1:
            res = self.__array + second_array
            print("\nResult of Addition:\n", res)
        elif op_type == 2:
            res = self.__array - second_array
            print("\nResult of Subtraction:\n", res)
        elif op_type == 3:
            res = self.__array * second_array
            print("\nResult of Multiplication:\n", res)
        elif op_type == 4:
            res = self.__array / second_array
            print("\nResult of Division:\n", res)

    # 3. Combine or Split
    def combine_array(self, second_elements):
        if not self.__has_array():
            return
        # Original shape mujab match karse (page 4 example pramane 2x3 + 2x3 = 4x3)
        second_array = np.array(second_elements).reshape(self.__array.shape)
        combined = np.vstack((self.__array, second_array))
        print("\nOriginal Array:\n", self.__array)
        print("Second Array:\n", second_array)
        print("\nCombined Array (Vertical Stack):\n", combined)

    # 4. Search, Sort, Filter
    def search_value(self, val):
        if not self.__has_array():
            return
        indices = np.where(self.__array == val)
        print(f"\nValue {val} found at indices:", indices)

    def sort_array(self):
        if not self.__has_array():
            return
        sorted_arr = np.sort(self.__array, axis=-1)
        print("\nOriginal Array:\n", self.__array)
        print("\nSorted Array:\n", sorted_arr)
        print("(Sorting applied row-wise.)")

    def filter_array(self, threshold):
        if not self.__has_array():
            return
        filtered = self.__array[self.__array > threshold]
        print(f"\nValues greater than {threshold}:\n", filtered)

    # 5. Aggregates and Statistics
    def compute_stats(self, option):
        if not self.__has_array():
            return
        print("\nOriginal Array:\n", self.__array)
        if option == 1:
            print("Sum of Array:", np.sum(self.__array))
        elif option == 2:
            print("Mean of Array:", np.mean(self.__array))
        elif option == 3:
            print("\nMedian of Array:", np.median(self.__array))
        elif option == 4:
            print("Standard Deviation:", np.std(self.__array))
        elif option == 5:
            print("Variance:", np.var(self.__array))


def main():
    analyzer = DataAnalytics()

    while True:
        print("\n=================================")
        print("Welcome to the NumPy Analyzer!")
        print("=================================")
        print("Choose an option:")
        print("1. Create a Numpy Array")
        print("2. Perform Mathematical Operations")
        print("3. Combine or Split Arrays")
        print("4. Search, Sort, or Filter Arrays")
        print("5. Compute Aggregates and Statistics")
        print("6. Exit")
        
        choice = input("Enter your choice: ").strip()

        if choice == '1':
            print("\nSelect the type of array to create:")
            print("1. 1D Array\n2. 2D Array\n3. 3D Array")
            dim = int(input("Enter your choice: "))
            
            if dim == 1:
                elements = DataAnalytics.parse_input(input("Enter elements separated by space: "))
                analyzer.create_array(1, None, elements)
            elif dim == 2:
                r = int(input("Enter the number of rows: "))
                c = int(input("Enter the number of columns: "))
                elements = DataAnalytics.parse_input(input(f"Enter {r*c} elements for the array separated by space: "))
                analyzer.create_array(2, (r, c), elements)
            elif dim == 3:
                d1 = int(input("Enter dimension 1: "))
                d2 = int(input("Enter dimension 2: "))
                d3 = int(input("Enter dimension 3: "))
                elements = DataAnalytics.parse_input(input(f"Enter {d1*d2*d3} elements separated by space: "))
                analyzer.create_array(3, (d1, d2, d3), elements)

            # Sub-menu indexing / slicing
            print("\nChoose an operation:")
            print("1. Indexing\n2. Slicing\n3. Go Back")
            sub_op = input("Enter your choice: ").strip()
            if sub_op == '2':
                r_range = input("Enter the row range (start:end): ").split(':')
                c_range = input("Enter the column range (start:end): ").split(':')
                analyzer.slice_array(int(r_range[0]), int(r_range[1]), int(c_range[0]), int(c_range[1]))

        elif choice == '2':
            if analyzer.get_array() is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose a mathematical operation:")
            print("1. Addition\n2. Subtraction\n3. Multiplication\n4. Division")
            op = int(input("Enter your choice: "))
            size = analyzer.get_array().size
            elements = DataAnalytics.parse_input(input(f"Enter the same-size array elements ({size} elements separated by space): "))
            analyzer.math_operation(op, elements)

        elif choice == '3':
            if analyzer.get_array() is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose an option:")
            print("1. Combine Arrays\n2. Split Array")
            ch = int(input("Enter your choice: "))
            if ch == 1:
                size = analyzer.get_array().size
                elements = DataAnalytics.parse_input(input(f"Enter the elements of another array to combine ({size} elements separated by space): "))
                analyzer.combine_array(elements)

        elif choice == '4':
            if analyzer.get_array() is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose an option:")
            print("1. Search a value\n2. Sort the array\n3. Filter values")
            ch = int(input("Enter your choice: "))
            if ch == 1:
                val = float(input("Enter value to search: "))
                analyzer.search_value(val)
            elif ch == 2:
                analyzer.sort_array()
            elif ch == 3:
                th = float(input("Enter threshold value for filter (> threshold): "))
                analyzer.filter_array(th)

        elif choice == '5':
            if analyzer.get_array() is None:
                print("\nPehle Array create karo!")
                continue
            print("\nChoose an aggregate/statistical operation:")
            print("1. Sum\n2. Mean\n3. Median\n4. Standard Deviation\n5. Variance")
            ch = int(input("Enter your choice: "))
            analyzer.compute_stats(ch)

        elif choice == '6':
            print("\nThank you for using the NumPy Analyzer! Goodbye!")
            break
        else:
            print("\nInvalid choice! Please try again.")

if __name__ == "__main__":
    main()
```
